```python
import pandas as pd
import numpy as np

# Create sample dataset for delivery robot simulation
data = {
    'delivery_id': [101, 102, 103, 104, 105, 106, 107, 108, 108, 109],
    'battery_level': [85, 45, 12, np.nan, 90, 30, 15, 60, 60, 95],
    'distance_km': [2.5, 8.0, 15.2, 4.1, np.nan, 12.0, 5.5, 3.2, 3.2, 1.0],
    'traffic_density': ['Low', 'High', 'Medium', 'Low', 'Medium', 'High', 'High', 'Low', 'Low', 'Medium'],
    'status': ['Completed', 'Completed', 'Failed', 'Completed', 'Pending', 'Failed', 'Completed', 'Completed', 'Completed', 'Completed']
}

df = pd.DataFrame(data)
print("Original Data:")
print(df.head())

# Data Preprocessing & Cleaning
df['battery_level'].fillna(df['battery_level'].mean(), inplace=True)
df['distance_km'].fillna(df['distance_km'].median(), inplace=True)
df.drop_duplicates(inplace=True)

# Feature Encoding and Heuristic Score Calculation
traffic_map = {'Low': 1, 'Medium': 2, 'High': 3}
df['traffic_code'] = df['traffic_density'].map(traffic_map)
df['heuristic_score'] = df['battery_level'] - (df['distance_km'] * df['traffic_code'])

print("\nProcessed Data:")
print(df[['delivery_id', 'battery_level', 'distance_km', 'traffic_density', 'heuristic_score']])
```
