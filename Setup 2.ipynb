```python
from datasets import load_dataset

# Download CoNaLa (curated version) from Hugging Face.
# The files are the Parquet copy that Hugging Face keeps of the dataset.
url = ("https://huggingface.co/datasets/neulab/conala/"
       "resolve/refs%2Fconvert%2Fparquet/curated/")
ds = load_dataset("parquet", data_files={
    "train": url + "train/0000.parquet",
    "test":  url + "test/0000.parquet",
})
print(ds)    # shows the train and test splits

train = ds["train"]    # 2,379 rows
test  = ds["test"]     # 500 rows (never train on these)

# The question is in the "intent" column, the answer in the "snippet" column


print("Question:", train[0]["intent"])
print("Answer: ", train[0]["snippet"])

# Take 10% of the TRAIN rows as the validation set (never from test)
split = train.train_test_split(test_size=0.1, seed=42)
train = split["train"]    # 2,141 rows: used for fine-tuning
val   = split["test"]     # 238 rows: the validation set
print("train:", len(train), " validation:", len(val), " test:", len(test))

# Optional: save the test set as a CSV file to look at in Excel
test.to_csv("conala_test.csv", index=False)
```
